```python
import pandas as pd
import polars as pl
import duckdb
import matplotlib.pyplot as plt
import plotly.express as px
import math

print("halo")
```

```python
df = pd.read_csv("AirBnb_data_case-study/Best_Property/listings_clean.csv")

# pastikan price numerik
df["price"] = (
    df["price"].astype(str)
    .str.replace("[$,]", "", regex=True)
    .str.strip()
    .astype(float)
)

# bikin histogram
plt.figure(figsize=(10,6))
plt.hist(df["price"], bins=100, color="skyblue", edgecolor="black")
plt.xlabel("Price (per night)")
plt.ylabel("Jumlah Listing (Airbnb)")
plt.title("Distribusi Price Airbnb")
plt.show()
```
