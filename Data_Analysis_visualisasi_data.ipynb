```python
import pandas as pd
import polars as pl
import duckdb
import matplotlib.pyplot as plt
import plotly.express as px

print("✅ pandas:", pd.__version__)
print("✅ polars:", pl.__version__)
print("✅ duckdb:", duckdb.__version__)

# Buat DataFrame contoh
df_pd = pd.DataFrame({
    "category": ["A", "B", "C", "A", "B"],
    "revenue": [100, 200, 150, 120, 180]
})

# --- Tes Pandas ---
print("\nPandas groupby:")
print(df_pd.groupby("category")["revenue"].sum())

# --- Tes Polars ---
df_pl = pl.from_pandas(df_pd)
print("\nPolars group_by:")
print(df_pl.group_by("category").agg(pl.col("revenue").sum()))

# --- Tes DuckDB (query DataFrame Pandas langsung) ---
print("\nDuckDB SQL query:")
print(duckdb.sql("SELECT category, SUM(revenue) AS rev FROM df_pd GROUP BY category").df())

# --- Tes Matplotlib ---
df_pd.groupby("category")["revenue"].sum().plot(kind="bar", title="Revenue by Category")
plt.show()

# --- Tes Plotly ---
fig = px.bar(df_pd, x="category", y="revenue", title="Revenue by Category (Interactive)")
fig.show()
```
