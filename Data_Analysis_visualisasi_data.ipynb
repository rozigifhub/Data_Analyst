```python
import pandas as pd
import polars as pl
import duckdb
import matplotlib.pyplot as plt
import plotly.express as px

data_frame1 = pd.read_csv("data/sales.csv")
data_frame2 = pd.read_csv("data/sales_update.csv")
# Buat tabel permanen dari DataFrame dengan alias "sales"
duckdb.sql("CREATE TABLE sales AS SELECT * FROM data_frame1")
duckdb.sql("CREATE TABLE sales_update AS SELECT * FROM data_frame2")
```
