# VISUALISASI DATA

```python
dataC = pd.DataFrame({
    "nama": ["Andi", "Budi", "Citra", "Dina"],
    "nilai": [85, 90, 78, 88],
    "usia": [22, 23, 21, 22]
})

plt.figure(figsize=(5,5))
plt.bar(dataC["nama"], dataC["nilai"], color="skyblue")
plt.title("Nilai Siswa")
plt.xlabel("Nama")
plt.ylabel("Nilai")
plt.show()

fig, ax = plt.subplots(figsize=(5, 5))
ax.bar(dataC["nama"], dataC["nilai"], color="skyblue")
ax.set_title("Nilai Siswa pakai OOP")
ax.set_xlabel("Nama")
ax.set_ylabel("Nilai")
plt.show()

plt.hist(dataC["nilai"], bins=20, color="orange")
plt.title("Distribusi Nilai")
plt.xlabel("Nilai")
plt.ylabel("Jumlah Siswa")
plt.show()
```

```python
sns.boxplot(x=dataC["nilai"])
plt.title("Boxplot Nilai Siswa")
plt.show()

plt.scatter(dataC["usia"], dataC["nilai"], color="green")
plt.title("Usia vs Nilai")
plt.xlabel("Usia")
plt.ylabel("Nilai")
plt.grid(True)
plt.show()
```

```python
##  Tujuan	                    Gunakan
### Bandingkan antar kategori	Bar chart
### Sebaran nilai/angka	        Histogram
### Outlier dan IQR	            Boxplot
### Hubungan dua variabel	    Scatter plot
```

```python
dataC = pd.DataFrame({
    "jurusan": ["Statistika", "Informatika", "Statistika", "Informatika"],
    "nilai": [85, 90, 78, 95]
})

# Hitung rata-rata nilai per jurusan
mean_per_jurusan = dataC.groupby("jurusan")["nilai"].mean()

# Plot bar chart rata-rata nilai per jurusan
fig, ax = plt.subplots(figsize=(5, 5))
ax.bar(mean_per_jurusan.index, mean_per_jurusan.values, color="skyblue")
ax.set_title("Rata-rata Nilai per Jurusan")
ax.set_xlabel("Jurusan")
ax.set_ylabel("Rata-rata Nilai")
plt.show()

# Boxplot nilai berdasarkan jurusan
sns.boxplot(x="jurusan", y="nilai", data=dataC)
plt.title("Boxplot Nilai per Jurusan")
plt.show()
```
